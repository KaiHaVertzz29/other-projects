```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Defining rolling function
def rolling(group,cols,new_cols):
    group=group.sort_values('Date')
    rolling=group[cols].rolling(3,closed='left').mean() #create rolling row
    group[new_cols]=rolling #assign row values to columns
    group=group.dropna(subset=new_cols)
    return(group)
```

```python
#column names and new_cols for rolling data
columns=['Result','xG','xGA','GF','GA','Poss']
new_cols=[f"{c}_rolling"for c in columns]
```

```python
epl_data=pd.read_csv('combined_data_1.csv')
#epl_team=pd.read_csv('team_2021.csv')
```

```python
epl_data.dtypes
```

```python
epl_data['Date']=pd.to_datetime(epl_data['Date'])
```

```python
#data preprocessing
epl_data['Time']=epl_data['Time'].str[:2].astype(int)
epl_data['GA'].fillna(0,inplace=True)
epl_data['GF'].fillna(0,inplace=True)
epl_data['Poss'].fillna(0,inplace=True)
epl_data['xG'].fillna(0,inplace=True)
epl_data['xGA'].fillna(0,inplace=True)
```

```python
#assigning rolling values to each team 
with_rolling=epl_data.groupby('Team').apply(lambda x: rolling(x,columns,new_cols))
```

```python
#dropping team level
with_rolling=with_rolling.droplevel('Team')
```

```python
#saving rolling table as csv for predictor module
with_rolling.to_csv('rolling_1.csv',index=False)
```
