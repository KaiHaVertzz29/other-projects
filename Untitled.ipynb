```python
import mysql.connector
import pandas as pd
import requests
```

```python
endpoint='https://v2.jokeapi.dev/joke/Any?amount=100 & blacklistFlags=racist,religious,explicit'

data=pd.DataFrame()

for i in range(10):
  response=requests.get(endpoint)
  response=response.json()['jokes']
  data=pd.concat([data,pd.DataFrame(response)],axis=0)
```

```python
flags_df=data['flags'].apply(pd.Series)
```

```python
flags_df.columns=('flags'+'_'+flags_df.columns).tolist()
```

```python
data.drop(columns='flags',inplace=True)
required_data=pd.concat([data,flags_df],axis=1)
```

```python
jokes_df=required_data.drop(columns=['flags_religious','flags_racist','flags_explicit','setup','delivery'])
jokes_df=jokes_df[['category','type','joke','flags_nsfw','flags_political','flags_sexist','safe','lang']]
```

```python
jokes_df=jokes_df.fillna('NULL')
```

```python
conn=mysql.connector.connect(
     host="127.0.0.1",
    port=3306,
    user='root',
    password='Dhruv_123',
    database='jokes_databse'
)
```

```python
cursor=conn.cursor()
```

```python
query="""
CREATE TABLE if not exists jokes_table (
    category varchar(20),
    type VARCHAR(10),
    joke varchar(1000),
    flags_nsfw boolean,
    flags_political boolean,
    flags_sexist boolean,
    safe boolean,
    lang varchar(5)
)
"""
```

```python
cursor.execute(query)
```

```python
for index,row in jokes_df.iterrows():
    insert_query="""
    Insert Into jokes_table (category, type , joke , flags_nsfw, flags_political, flags_sexist,safe, lang)
    Values (%s,%s,%s,%s,%s,%s,%s,%s)
    """
    cursor.execute(insert_query,(row['category'],row['type'],row['joke']
                                 ,int(row['flags_nsfw']),int(row['flags_political'])
                                 ,int(row['flags_sexist']),int(row['safe']),row['lang']))
conn.commit()
```
